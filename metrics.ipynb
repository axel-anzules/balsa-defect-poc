```python
from ultralytics import YOLO

model = YOLO("C:\\Users\\ASUS_TUF_I7\\OneDrive\\Documentos\\Docs CV Balsas Defects Project\\balsa-defect-poc\\exp-5.pt")

results = model.val(
    data="C:\\Users\\ASUS_TUF_I7\\OneDrive\\Documentos\\Docs CV Balsas Defects Project\\balsa-defect-poc\\dataset\\poc-defects-on-wood-balsa-panelsv15-version-with-pre-6c5d496f\\data.yaml",
    plots=True
)

class_names = model.names

print(f"mAP50: {results.box.map50:.4f}")
print(f"mAP50-95: {results.box.map:.4f}")
print(f"Precision: {results.box.mp:.4f}")
print(f"Recall: {results.box.mr:.4f}")

for i, map_value in enumerate(results.box.maps):
    print(f"{class_names[i]}: {map_value:.4f}")
```

```python
for i, name in model.names.items():
    print(name)
    print(results.class_result(i))
```

```python
import pandas as pd
rows = []
for i, name in model.names.items():
    precision, recall, map50, map50_95 = results.class_result(i)
    rows.append({
        "Class": name,
        "Precision": precision,
        "Recall": recall,
        "mAP50": map50,
        "mAP50-95": map50_95
    })
df = pd.DataFrame(rows)
print(df)
```
